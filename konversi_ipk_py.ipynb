```python
studentData     = input("Masukkan Data Mahasiswa (Nama, NIM, Tugas, UTS, UAS): ")
dataParts       = studentData.split(",")
studentName     = dataParts[0].strip()
studentId       = dataParts[1].strip()
assignmentScore = float(dataParts[2].strip())
midtermScore    = float(dataParts[3].strip())
finalExamScore  = float(dataParts[4].strip())
finalScore      = (assignmentScore * 0.20) + (midtermScore * 0.35) + (finalExamScore * 0.45)

print("\n" + "+" + "-" * 38 + "+")
print(f"| {'TRANSKRIP NILAI AKADEMIK':^36} |")
print("+" + "-" * 38 + "+")
print(f"| {'Nama':<15}: {studentName:<20} |")
print(f"| {'NIM':<15}: {studentId:<20} |")
print("+" + "-" * 38 + "+")
print(f"| {'Nilai Tugas':<15}: {assignmentScore:>20.1f} |")
print(f"| {'Nilai UTS':<15}: {midtermScore:>20.1f} |")
print(f"| {'Nilai UAS':<15}: {finalExamScore:>20.1f} |")
print("+" + "-" * 38 + "+")
print(f"| {'Nilai Akhir':<15}: {finalScore:>20.1f} |")
print("+" + "-" * 38 + "+")
```
