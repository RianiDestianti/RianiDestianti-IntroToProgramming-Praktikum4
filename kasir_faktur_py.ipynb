```python
customerName        = input("Masukkan Nama Pelanggan: ")
totalBeforeDiscount = float(input("Masukkan Total Belanja Sebelum Diskon: Rp"))
discountPercentage  = float(input("Masukkan Persentase Diskon (%): "))

discountAmount     = totalBeforeDiscount * discountPercentage / 100
totalAfterDiscount = totalBeforeDiscount - discountAmount

print("\n" + "=" * 36)
print(f"{'NOTA PEMBAYARAN':^36}")
print(f"{customerName:^36}")
print("=" * 36)

print(f"{'Total Awal':<20}{'Rp' + f'{totalBeforeDiscount:,.2f}':>16}")
print(f"{f'Diskon ({discountPercentage:g}%)':<20}{'Rp' + f'{discountAmount:,.2f}':>16}")
print("-" * 36)
print(f"{'Total Bayar':<20}{'Rp' + f'{totalAfterDiscount:,.2f}':>16}")

print("=" * 36)
print(f"{'Terima Kasih Telah Berbelanja!':^36}")
print("=" * 36)
```
